```python
import torch
from torch import nn

# 1. Implémentation manuelle de la corrélation croisée 2D [cite: 1065, 1087, 1094]
def corr2d(X, K):
    h, w = K.shape
    Y = torch.zeros((X.shape[0] - h + 1, X.shape[1] - w + 1))
    for i in range(Y.shape[0]):
        for j in range(Y.shape[1]):
            Y[i, j] = (X[i:i+h, j:j+w] * K).sum()
    return Y

# 2. Architecture CNN (Inspirée de LeNet) [cite: 1419, 1420, 1422, 1426, 1427]
class LeNetFashion(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.net = nn.Sequential(
            # Bloc 1 : Convolution + Activation + Pooling [cite: 1420, 1421, 1422]
            nn.LazyConv2d(6, kernel_size=5, padding=2),
            nn.ReLU(), # Remplacement de Sigmoid par ReLU (meilleure pratique moderne)
            nn.MaxPool2d(kernel_size=2, stride=2), 
            
            # Bloc 2 [cite: 1423, 1424, 1425]
            nn.LazyConv2d(16, kernel_size=5),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),
            
            # Classifieur dense [cite: 1426, 1427, 1429, 1431]
            nn.Flatten(),
            nn.LazyLinear(120),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.LazyLinear(84),
            nn.ReLU(),
            nn.LazyLinear(num_classes)
        )

    def forward(self, x):
        return self.net(x)

# 3. Inspection des dimensions (Très utile pour le rapport) [cite: 1446, 1449, 1450]
# model_cnn = LeNetFashion()
# X_dummy = torch.randn(1, 1, 28, 28) # Format Fashion-MNIST
# for layer in model_cnn.net:
#     X_dummy = layer(X_dummy)
#     print(layer.__class__.__name__, X_dummy.shape)
```
