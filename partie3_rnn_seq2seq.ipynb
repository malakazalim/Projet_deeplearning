```python
import torch
from torch import nn

# 1. Comparaison des cellules récurrentes de base [cite: 529, 594, 623]
# Ces modules s'attendent à une entrée de forme (batch_size, sequence_length, features)
rnn_simple = nn.RNN(input_size=32, hidden_size=64, batch_first=True)
lstm_cell = nn.LSTM(input_size=32, hidden_size=64, batch_first=True)
gru_cell = nn.GRU(input_size=32, hidden_size=64, batch_first=True)

# 2. Squelette du modèle Seq2Seq [cite: 769, 770, 776, 777]
class Seq2Seq(nn.Module):
    def __init__(self, encoder, decoder):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder

    def forward(self, src, tgt_in):
        # L'encodeur produit les états cachés finaux [cite: 776]
        enc_outputs, enc_state = self.encoder(src)
        
        # Le décodeur utilise ces états pour générer la séquence [cite: 777]
        dec_outputs, _ = self.decoder(tgt_in, enc_state, enc_outputs)
        return dec_outputs

# 3. Exemple de mécanique d'entraînement avec Gradient Clipping (BPTT) [cite: 562, 564, 781]
# loss.backward()
# torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
# optimizer.step()
# optimizer.zero_grad()
```
